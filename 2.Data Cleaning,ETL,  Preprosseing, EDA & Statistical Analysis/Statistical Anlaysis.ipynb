    Objective, Dataset Scope, and End-to-End ETL + Ensemble Regression Pipeline
  </p>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color : black">
    The primary objective of this work is to develop a <b>robust ETL-driven ensemble regression framework</b> for predicting
    <b>methanol and ethanol yields</b> from <b>electrochemical methane (CH₄) conversion</b> experiments.
    The pipeline integrates automated <b>data preprocessing, feature engineering, exploratory analysis, and ensemble machine learning</b>
    to support catalyst performance optimization under varying operating conditions.
    The dataset is <b>electrolyzer-centered</b>, while also incorporating complementary data from
    <b>electrochemical devices</b> and <b>solar-assisted systems</b> to improve model generalization and robustness.
  </p>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color : green ; font-weight : bold">
    Data Engineering & ETL Pipeline (Regression-Focused)
  </p>
  <ul>
    <li>Designed an end-to-end <b>ETL pipeline</b> for automated ingestion, cleaning, transformation, and preparation of experimental datasets.</li>
    <li>Unified data from multiple experimental platforms (electrolyzer, electrochemical device, solar-assisted systems).</li>
    <li>Standardized device descriptors to reduce category noise and improve regression stability.</li>
    <li>Generated a clean, reproducible, and ML-ready dataset optimized for ensemble regression modeling.</li>
  </ul>

  <p><b>Data Preprocessing & Feature Engineering</b></p>
  <ul>
    <li>Standardized missing values (<i>NANE</i>, <i>NONE</i>, blanks) and corrected mixed numeric/string columns.</li>
    <li>Resolved unit inconsistencies (e.g., light intensity, voltage, current density).</li>
    <li>Cleaned categorical labels to ensure consistent feature representation.</li>
    <li>Applied <b>normalization and scaling</b> to numerical variables.</li>
    <li>Handled sparsity in catalyst composition features using aggregation and structured grouping (e.g., total metal loading).</li>
    <li>Removed redundant and low-information columns to enhance model interpretability.</li>
    <li>Detected and removed extreme outliers caused by experimental noise (e.g., unrealistic voltage ranges).</li>
    <li>Performed <b>correlation analysis</b> and variance-based filtering to reduce multicollinearity.</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color : green ; font-weight : bold">
    Dataset Design
  </p>
  <p style="text-align: justify; font-family: cursive; font-size: 14px; color : black">
    <b>Input Features:</b> Catalyst composition (Pt at%, Co wt.%, TiO₂ wt.%, g-C₃N₄ wt.%, metal ratios),
    reaction conditions (voltage, reaction time, CH₄ flow rate, electrolyte concentration, current density),
    device indicators (Electrolyzer / Electrochemical device / Solar light),
    and material characterization parameters (e.g., BET surface area).
  </p>
  <p style="text-align: justify; font-family: cursive; font-size: 14px; color : black">
    <b>Target Variables (Regression):</b> Methanol yield (μmol/ml) and ethanol yield (μmol/ml).
  </p>

    Machine Learning Tasks (Regression Only)
  </p>
  <ul>
    <li><b>Primary Task:</b> Predict methanol yield from catalyst composition and electrochemical operating conditions.</li>
    <li><b>Secondary Task:</b> Predict ethanol yield using the same feature space to compare product selectivity.</li>
    <li>Applied <b>k-fold cross-validation</b> to ensure generalization across diverse experimental conditions.</li>

#### ANOVA Analysis

```python
anova_df = df[['Electrolyte', 'Methanol Yield (µmol/ml )']].dropna()

groups = [
    g['Methanol Yield (µmol/ml )'].values
    for _, g in anova_df.groupby('Electrolyte')
]

F_stat, p_val = stats.f_oneway(*groups)

print(f"F-statistic: {F_stat:.3f}")
print(f"P-value: {p_val:.4e}")
```

```python
anova_df = df[['Process type', 'Ethanol Yield (µmol/ml )']].dropna()

groups = [
    g['Ethanol Yield (µmol/ml )'].values
    for _, g in anova_df.groupby('Process type')
]

F_stat, p_val = stats.f_oneway(*groups)

print(f"F-statistic: {F_stat:.3f}")
print(f"P-value: {p_val:.4e}")
```
