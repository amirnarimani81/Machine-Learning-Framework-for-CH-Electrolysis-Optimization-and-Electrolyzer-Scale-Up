```python
sql= """ SELECT *
FROM preprocessed
WHERE  Methanol_Yield_micromol_ml_ IS NOT NULL"""

df = pd.read_sql(sql, engine)
df.head()
```

# Average Yield per Catalyst Combination

```python
sql= """ SELECT 
    cathod_catalyst,
    anode_catalyst,
    AVG(Methanol_Yield_micromol_ml_) AS avg_yield,
    COUNT(*) AS experiment_count
FROM preprocessed
GROUP BY cathod_catalyst, anode_catalyst
ORDER BY avg_yield DESC """

df = pd.read_sql(sql, engine)
df.head()
```

# Yield Efficiency (Yield per Reaction Time)

```python
sql= """ SELECT 
    cathod_catalyst,
    Methanol_Yield_micromol_ml_ / Reaction_Time_hr AS efficiency FROM preprocessed """

df=pd.read_sql(sql, engine)
df.head()
```

# Detect Duplicate Experimental Conditions

```python
sql=""" SELECT 
    cathod_catalyst,
    anode_catalyst,
     Voltage_V,
    CH4_Flow_rate_sscm,
    COUNT(*) AS duplicates
FROM preprocessed
GROUP BY 
    cathod_catalyst,
    anode_catalyst,
     Voltage_V,
    CH4_Flow_rate_sscm
HAVING COUNT(*) > 1"""

df = pd.read_sql(sql, engine)
df.head()
```

# Catalyst Metal Composition Summary

```python
sql=""" SELECT
    process_type,
    AVG(Pt_atpercent_in_metal_element) AS avg_pt,
    AVG(Ni_atpercent_in_metal_element) AS avg_ni,
    AVG(Co_atpercent_in_metal_element) AS avg_pd
FROM preprocessed
GROUP BY process_type"""

df = pd.read_sql(sql, engine)
df.head()
```
